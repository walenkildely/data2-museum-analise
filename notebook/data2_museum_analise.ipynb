 Como a composição de tipos de museu varia geograficamente

```python
# Cruza Estado (Administrative Location) x Museum Type, contando quantos
# museus de cada tipo existem em cada estado
composicao_geografica = pd.crosstab(
    df_limpo["State (Administrative Location)"],
    df_limpo["Museum Type"]
)

# Converte contagem absoluta em PROPORCAO (%) dentro de cada estado
# normalize="index" faz cada LINHA (cada estado) somar 100%
composicao_percentual = pd.crosstab(
    df_limpo["State (Administrative Location)"],
    df_limpo["Museum Type"],
    normalize="index"
) * 100

# Quantos museus tem cada estado no total (para saber se a base de
# comparacao eh confiavel -- estado com poucos museus tem % instavel)
total_por_estado = df_limpo["State (Administrative Location)"].value_counts()

print(total_por_estado.head(10))
print(total_por_estado.tail(10))
```

verificando composição nacional

```python
# Composição nacional (média geral) de tipos de museu, em %
composicao_nacional = df_limpo["Museum Type"].value_counts(normalize=True) * 100

# Diferença entre cada estado e a média nacional, por tipo de museu
desvio_estado = composicao_percentual.subtract(composicao_nacional, axis=1)

# Maior desvio absoluto de cada estado (qual estado mais "destoa" do padrão)
maior_desvio_por_estado = desvio_estado.abs().max(axis=1).sort_values(ascending=False)

print(maior_desvio_por_estado.head(10))
```

verificando qual categoria puxa o desvio de cada estado

```python
# Para os 3 estados de maior desvio, qual tipo de museu causa o desvio
for estado in ["UT", "DC", "NH"]:
    print(estado, "->", desvio_estado.loc[estado].abs().idxmax(),
          round(desvio_estado.loc[estado].abs().max(), 1), "p.p.")
```

Conclusão: composição geográfica por tipo de museu

A composição de tipos de museu varia significativamente entre estados
quando comparada à média nacional. Os estados com maior desvio são
Utah (24,3 p.p.), Distrito de Columbia (23,8 p.p.) e New Hampshire
(20,6 p.p.) — e nos três casos o desvio vem da mesma categoria:
HISTORIC PRESERVATION, presente nesses estados em proporção bem
acima da média nacional (que já é alta, ~55%, conforme RF03-a).
Isso sugere que o setor de "preservação histórica" não está
distribuído de forma uniforme pelo país: ele se concentra ainda mais
em alguns estados específicos, reforçando o achado de RF03-a de que
qualquer leitura de "mercado de museus" precisa ser segmentada — por
tipo e, agora, também por região — para não mascarar essas
concentrações.

**Discrepâncias que distorcem a leitura do mercado**

Sim, foram identificadas duas discrepâncias relevantes durante a análise:

**1. Duplicação institucional (Padrão 2).** Organizações-mãe (ex.:
universidades como Harvard e Yale) declaram receita fiscal uma única
vez, mas o dataset replica esse valor em múltiplas linhas-filha
(museus/coleções pertencentes a elas). Identificamos 905 grupos
afetados (1.125 linhas, ~5% da base financeira). Sem tratamento, isso
inflava a média de Revenue em até 89% em algumas categorias (Natural
History Museum), pois o mesmo dinheiro era contado várias vezes sob
rótulos de tipo diferentes.

**2. Concentração de mercado por tipo e por região.** HISTORIC
PRESERVATION representa 55% da base financeira e, além disso, está
desproporcionalmente concentrada em estados como Utah, DC e New
Hampshire. Isso significa que qualquer leitura agregada de
"receita média do setor" sem segmentar por tipo e por estado mistura
um perfil de mercado muito específico (preservação histórica, receita
mediana próxima de zero) com outros perfis bem diferentes (ex.: Art
Museum, receita mediana mais alta).

Ambas as discrepâncias foram tratadas no notebook (Padrão 2 com regra
de deduplicação de receita por grupo; concentração regional registrada
como achado, não removida — é um fato do mercado, não um erro de dado).

# Recomendações estratégicas finais

**1. Posicionar o Nur Mah Museum aproveitando o perfil de receita do
seu próprio tipo .**
Evidência: o boxplot mostra Art Museum com a mediana de
Revenue mais alta entre os 9 tipos analisados . Isso
indica que museus de arte, como categoria, operam com receita típica
acima da maioria do setor — um argumento de benchmarking favorável
para captação de recursos e parcerias.

**2. Evitar comparações de "receita média do setor" sem segmentação
por tipo e por região.**
Evidência: a duplicação institucional do Padrão 2.
inflou médias em até 89% antes do tratamento, e HISTORIC PRESERVATION
concentra 55% da base nacionalmente e ainda mais em estados como UT,
DC e NH . Qualquer benchmark de mercado usado pela
diretoria deve sempre segmentar por Museum Type e, quando relevante,
por estado — média agregada sem esse filtro é estatisticamente
enganosa para a tomada de decisão.

**3. Priorizar parcerias/benchmarking com organizações do mesmo porte
mediano, não com outliers extremos.**
Evidência: mesmo após o tratamento do Padrão 2, outliers residuais
persistem em todas as categorias no boxplot — são casos
reais de museus muito grandes, não representativos da maioria do
setor. Usar esses casos extremos como referência de "padrão de
mercado" levaria a metas de captação irreais; a mediana por tipo é a
referência mais robusta para metas e comparações.
