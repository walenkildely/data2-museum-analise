Verificando duplicatas por nome (nome + cidade + estado)

Verificando padrão de duplicatas por nome (nome + cidade + estado), ordenando a partir da coluna (Museum Name) e exibindo apenas colunas especificas

```python
df[df.duplicated(subset=["Museum Name", "City (Administrative Location)", "State (Administrative Location)"], keep=False)].sort_values("Museum Name")[["Museum Name", "Legal Name", "Institution Name", "City (Administrative Location)", "State (Administrative Location)", "Revenue"]]
```

Adicionando o Legal Name na verificação de duplicadas

```python
df.duplicated(subset=["Museum Name", "Legal Name", "City (Administrative Location)", "State (Administrative Location)"]).sum()
```

Verificando se o 0 é informado ou não informado (valor real declarado)

```python
df[df["Revenue"] == 0][["Tax Period", "Income"]].isnull().sum()
```

Contando campos preenchidos por linha

```python
df["completude"] = df.notna().sum(axis=1)
```

Fazendo analise da coluna completude

```python
df["completude"].describe()
```

Isolando os duplicados

```python
duplicados = df[df.duplicated(subset=["Museum Name", "Legal Name", "City (Administrative Location)", "State (Administrative Location)"], keep=False)]
```

Fazendo Analise dos duplicados

```python
duplicados["completude"].describe()
```
