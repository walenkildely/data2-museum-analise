```python
df_financeiro = df_limpo.dropna(subset=["Revenue"])
df_financeiro.shape
```

```python
df_financeiro[ df_financeiro["Revenue"] == 5840349457 ][["Museum Name", "Museum Type", "Legal Name"]]
```

Verificando duplicatas de Revenue

```python
contagem = df_financeiro.groupby(["Legal Name", "Revenue"]).size()
contagem[(contagem > 1) & (contagem.index.get_level_values("Revenue") != 0)]
```

identificando ocorrencias

```python
df_financeiro = df_financeiro.copy()
df_financeiro["pos_no_grupo"] = df_financeiro.groupby(["Legal Name", "Revenue"]).cumcount()
mascara_padrao2 = (df_financeiro["pos_no_grupo"] > 0) & (df_financeiro["Revenue"] != 0)
print("Linhas tratadas como Revenue duplicado (Padrão 2):", mascara_padrao2.sum())
df_financeiro_corrigido = df_financeiro.copy()
df_financeiro_corrigido.loc[mascara_padrao2, "Revenue"] = pd.NA
```

removendo coluna auxiliar

```python
df_financeiro_corrigido = df_financeiro_corrigido.drop(columns=["pos_no_grupo"])
df_financeiro.drop(columns=["pos_no_grupo"], inplace=True)
```

confirmação final

```python
print(df_financeiro_corrigido.shape)
print(df_financeiro_corrigido["Revenue"].isna().sum(), "linhas agora sem Revenue (excluídas das contas financeiras, mas mantidas na base)")
```
